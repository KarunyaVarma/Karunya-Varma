```python
import numpy as np
import pandas as pd
import tensorflow as tf
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split

from sklearn.preprocessing import MinMaxScaler
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score


# -------------------------------------------------
# Input data & Reproducibility
# -------------------------------------------------

tf.keras.utils.set_random_seed(42)

# Read the CSV file uploaded to Google Colab
data = pd.read_csv('/content/GTPP-DATASET.csv')

print("Data Preview:")
print(data.head())

# 8 input features
X = data[[
    'AT',
    'AP',
    'DF',
    'CH',
    'GP',
    'CPR',
    'CPD',
    'TTXM'
]].values

# Target Output
y = data['EP'].values

print("\nInput shape:", X.shape)
print("Output shape:", y.shape)


# -------------------------------------------------
# Train-Test Split (80:20 Ratio)
# -------------------------------------------------

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)


# -------------------------------------------------
# MinMax Scaling
# -------------------------------------------------

# CHANGED: Fit MinMaxScaler on training set, then transform both train and test
scaler = MinMaxScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)


# -------------------------------------------------
# Deep Neural Network Architecture
# -------------------------------------------------

model = tf.keras.models.Sequential([

    # Input layer: 8 inputs (AT, AP, DF, CH, GP, CPR, CPD, TTXM)
    tf.keras.layers.Input(shape=(8,)),

    # Hidden layer 1
    tf.keras.layers.Dense(1500, activation='relu'),

    # Output layer
    tf.keras.layers.Dense(1)
])

model.summary()


# -------------------------------------------------
# Compile Neural Network
# -------------------------------------------------

model.compile(
    optimizer=tf.keras.optimizers.Adam(
        learning_rate=0.001,
        beta_1=0.9,
        beta_2=0.999
    ),

    loss='mse',

    metrics=['mae']
)


# -------------------------------------------------
# Train Neural Network
# -------------------------------------------------

history = model.fit(
    X_train_scaled,
    y_train,
    epochs=50,
    batch_size=32,
    validation_split=0.2,

)


# -------------------------------------------------
# Evaluation & Full Metric Calculation
# -------------------------------------------------

# Predict on test set
y_pred = model.predict(X_test_scaled).flatten()

# CHANGED: Calculate all paper metrics (MAE, MSE, RMSE, R2)
mae = mean_absolute_error(y_test, y_pred)
mse = mean_squared_error(y_test, y_pred)
rmse = np.sqrt(mse)
r2 = r2_score(y_test, y_pred)


print("ANN MODEL TEST METRICS:-     ")
print("\nTest MAE:", mae)
print("\nTest MSE:", mse)
print("\nTest RMSE:", rmse)
print("\nTest R2 Score:", r2)
```
